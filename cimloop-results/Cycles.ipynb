```python
import os
import re

def extract_layer_number(folder_name):
    match = re.search(r'layer_(\d+)', folder_name)
    return int(match.group(1)) if match else float('inf')

def extract_cycles(model, tech_type, size):
    output_dir = f"../outputs/{tech_type}_{model}_{size}_cycle9/"
    max_cycle = 0
    folders = sorted([folder for folder in os.listdir(output_dir) if model in folder], key=extract_layer_number)

    for folder in folders:
        if model in folder:
            # print(f"[{tech_type}/{model}_{size}]")
            if os.path.isdir(output_dir + folder):
                stat_file = output_dir + folder + "/timeloop-mapper.stats.txt"
                
                with open(stat_file, 'r') as f:
                    cycle_stat = [line for line in f if line.startswith("Cycles:")][0]                  
                    cycle = float(cycle_stat.split()[1]) 
                    max_cycle = max(max_cycle, cycle)
    # print(f"Max cycle: {max_cycle}")
    return max_cycle      

models = ["alexnet", "vgg16", "resnet18"]
tech_types = ["SRAM", "RRAM"]
size = "128"

for model in models:
    for tech in tech_types:
        print(f"[{tech}/{model}_{size}]: Max cycle: {extract_cycles(model, tech, size)}")
        # extract_cycles(model, tech, size)
```
